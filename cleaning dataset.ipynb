```python
import pandas as pd
# STEP 1: Load the dataset
print("--- Step 1: Loading Data ---")
file_path = "/content/superstore.csv"
df = pd.read_csv(file_path)
print("Dataset loaded successfully! Here is a preview of the first 5 rows:")
print(df.head())
# STEP 2: Handle Missing Values
print("\n--- Step 2: Checking for Missing Values ---")
print("Missing values per column before fix:")
print(df.isnull().sum())
df = df.fillna(0)
print("\nMissing values have been handled!")
# STEP 3: Drop Duplicate Rows
print("\n--- Step 3: Removing Duplicates ---")
duplicate_count = df.duplicated().sum()
print(f"Found {duplicate_count} duplicate rows.")
if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicates removed successfully!")
# STEP 4: Convert Data Types (e.g., Strings to Dates)
print("\n--- Step 4: Converting Data Types ---")
date_columns = ['Order Date', 'Ship Date', 'date', 'Date']
for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')
        print(f"Successfully converted '{col}' to datetime format.")
print("\nFinal column data types:")
print(df.dtypes)
# STEP 5: Export Cleaned Data to a New CSV
print("\n--- Step 5: Exporting Cleaned Data ---")
output_filename = "Cleaned_Dataset.csv"
df.to_csv(output_filename, index=False)
print(f"🎉 Success! Your cleaned dataset has been saved as: {output_filename}")
print("You can find and download this file from the left sidebar folder icon.")




```
